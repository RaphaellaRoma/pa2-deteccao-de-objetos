# Inferência — identidade e contagem em uma sequência MOT17

Informe o caminho da sequência (pasta `MOT17-XX-FRCNN`). Este notebook carrega o checkpoint treinado, executa o rastreador temporal sem treino e grava um MP4 com cores consistentes por ID, além das predições em formato MOT e da contagem de identidades únicas. O vídeo pode conter sobrecontagem quando uma pessoa recebe IDs diferentes após uma falha de associação.

```python
from pathlib import Path
import os, sys
from IPython.display import Video, display
REPO = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p/"configs/parte1.json").exists()), None)
if REPO is None:
    raise RuntimeError("Abra o notebook dentro do repositório.")
os.chdir(REPO)
if str(REPO) not in sys.path:
    sys.path.insert(0, str(REPO))
from render_video import render

SEQUENCE_DIR = Path("data/MOT17/train/MOT17-09-FRCNN")
CHECKPOINT = Path("checkpoints/temporal_mot17.pt")
OUTPUT = Path("outputs/inferencia/MOT17-09.mp4")
DEVICE = "auto"  # CUDA quando disponível
MAX_FRAMES = None  # duração inteira; número positivo apenas para ensaio rápido
```

```python
result = render(SEQUENCE_DIR, CHECKPOINT, OUTPUT, DEVICE, max_frames=MAX_FRAMES)
print("Quadros:", result["frames"])
print("Identidades únicas previstas:", result["unique_identities"])
print("Vídeo:", result["video"])
print("Predições MOT:", result["predictions"])
display(Video(result["video"], embed=True))
```

O checkpoint é gerado pelo comando `python run_temporal.py all --device cuda`. O vídeo usa a fonte pública FRCNN congelada e o mesmo corte de confiança e limiar de associação da integração. O GT não participa da inferência.
